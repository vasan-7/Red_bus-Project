```python

from selenium import webdriver
from selenium.webdriver import ActionChains
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd

driver = webdriver.Chrome()
driver.maximize_window()

all_buses_data = []

for i,r in df.iterrows():
    link=r["route_link"]
    route=r["route_name"]

    # Loop through each link
    driver.get(link)
    time.sleep(3)
    
    
    try:
        button = driver.find_elements(By.XPATH, "//button[contains(@class,'primaryButton___3262c2') and contains(@class,'withAnimation___369432')]")
        button[0].click()
        time.sleep(3)
    except Exception as e:
        print("No button found or unable to click:", e)
        

    bus_names = driver.find_elements(By.XPATH, "//div[@class='travelsName___950ec8']")
    bus_types = driver.find_elements(By.XPATH, "//div[@class='bottomRow___0a77c4']")
    departing_time = driver.find_elements(By.CLASS_NAME, "boardingTime___a78ae0")
    duration = driver.find_elements(By.CLASS_NAME, "duration___b3a515")
    reaching_time= driver.find_elements(By.CLASS_NAME, "droppingTime___c814da")
    star_ratings = driver.find_elements(By.XPATH, "//div[@class='rating___24df2f']")
    prices = driver.find_elements(By.XPATH, "//div[@class='fareWrapper___e79a42']")
    seats_availabilities = driver.find_elements(By.XPATH, "//div[@class='seatsWrap___fa5a7a']")
    #print(bus_name,bus_type,departing_time,duration_time,reaching_time,star_rating,price,seats_availability)


    for idx in range(len(bus_names)):
        bus_data = {
            "route": route,
            "link": link,
            "Bus Name": bus_names[idx].text,
            "Bus Type": bus_types[idx].text if idx < len(bus_types) else "N/A",
            "Departing Time": departing_time[idx].text if idx < len(departing_time) else "N/A",
            "Duration Time": duration[idx].text if idx < len(duration) else "N/A",
            "Reaching Time": reaching_time[idx].text if idx < len(reaching_time) else "N/A",
            "Star Rating": star_ratings[idx].text if idx < len(star_ratings) else "N/A",
            "Price": prices[idx].text if idx < len(prices) else "N/A",
            "Seats Availability": seats_availabilities[idx].text if idx < len(seats_availabilities) else "N/A",
        }
        all_buses_data.append(bus_data)
        print(bus_data)

    
```
